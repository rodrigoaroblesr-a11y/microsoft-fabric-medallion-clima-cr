```python
%%sql
-- CREACIÓN DE LA CAPA ORO (GOLD): MÉTRICAS EJECUTIVAS POR PROVINCIA
CREATE OR REPLACE TABLE lk_costarica_clima.dbo.oro_resumen_provincial
AS
SELECT 
    provincia_nombre,
    
    -- TRANSFORMACIÓN DE AGREGACIÓN: Calculamos resúmenes globales por provincia
    COUNT(canton_nombre) AS total_cantones_monitoreados,
    ROUND(AVG(temperatura_celsius), 1) AS promedio_temperatura_celsius,
    ROUND(SUM(precipitacion_mm), 2) AS total_lluvia_acumulada_mm,
    
    -- Contamos cuántas alertas críticas están activas en este momento por provincia
    SUM(CASE WHEN estado_clima = 'Alerta de Tormenta' THEN 1 ELSE 0 END) AS cantidad_alertas_tormenta,
    SUM(CASE WHEN estado_clima = 'Calor Extremo' THEN 1 ELSE 0 END) AS cantidad_alertas_calor

FROM lk_costarica_clima.dbo.plata_monitoreo_distrital
GROUP BY provincia_nombre;
```
