```python
%%sql
-- ACTUALIZACIÓN CON TRANSFORMACIÓN DE TIPO DE DATO (CAST A DECIMAL)
CREATE OR REPLACE TABLE lk_costarica_clima.dbo.plata_monitoreo_distrital
AS
SELECT 
    c.canton_id,
    c.fecha_registro,
    g.provincia_nombre,
    g.canton_nombre,
    
    -- TRANSFORMACIÓN: Forzamos el tipo DECIMAL para eliminar el ruido de flotantes
    -- DECIMAL(4,1) significa: 4 dígitos en total y 1 de ellos es decimal (ej. 32.8)
    -- DECIMAL(5,2) significa: 5 dígitos en total y 2 de ellos son decimales (ej. 11.30)
    CAST(c.temperatura_celsius AS DECIMAL(4,1)) AS temperatura_celsius,
    CAST(c.precipitacion_mm AS DECIMAL(5,2)) AS precipitacion_mm,
    CAST(c.humedad_porcentaje AS INT) AS humedad_porcentaje,
    
    -- Regla de Negocio
    CASE 
        WHEN c.precipitacion_mm >= 35.0 AND c.temperatura_celsius <= 22.0 THEN 'Alerta de Tormenta'
        WHEN c.temperatura_celsius >= 30.0 THEN 'Calor Extremo'
        ELSE 'Normal'
    END AS estado_clima

FROM lk_costarica_clima.dbo.bronce_meteo_lecturas c
INNER JOIN lk_costarica_clima.dbo.bronce_geo_cantones g 
    ON c.canton_id = g.canton_id;
```
